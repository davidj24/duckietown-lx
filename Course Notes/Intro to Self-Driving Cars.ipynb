### <p style="text-align:center; color:white">The Stateful Approach</p>

<div style="border: 2px solid white; padding: 15px; width: 80%; margin: auto; text-align: center; background-color: black; color:white">
    
This is the approach we’re using!!!!

In Stateful architectures, we give the machine a “belief” or state about the world allowing it to make decisions
The perception block is to receives observations and forms a state about the robot’s surroundings
The planner creates a plan to act upon using the belief
The control block finally takes the plan and creates commands to execute the plan
1. The process begins with observations taken from the camera that go to the perception block to create states, which are filtered versions of observations that the model can use in its planner block

2. The planner block takes those states and creates a best course of action
3. This course of action is passed to control which will create the commands needed to follow that course
4. The commands are given to actuators that make the robot execute those commands


When building an architecture for the car, we can start off with the above shown generic structure and then make it more complex when needed

In Duckiebot, given a stream of images, the most frequently running loop is the one that computes the position in the line and the control commands
Self driving cars require an attention mechanism so that all the computing power is used for only the relevant parts of the robot’s surroundings

The approach that we will be using will be using a deep neural network, preferably a CNN trained on simulation data, to translate data from the camera to directly into commands for the car’s wheels

</div>
